```python
import mysql.connector

con = mysql.connector.connect(
    host="localhost",
    user="root",
    password="root",
    database="car_rental"
)

cur = con.cursor()

while True:
    print("\n===== CAR RENTAL MANAGEMENT SYSTEM =====")
    print("1. Display Cars")
    print("2. Display Customers")
    print("3. Display Bookings")
    print("4. Search Car")
    print("5. Book a Car")
    print("6. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        cur.execute("SELECT * FROM cars")
        data = cur.fetchall()

        for row in data:
            print(row)

    elif choice == 2:
        cur.execute("SELECT * FROM customers")
        data = cur.fetchall()

        for row in data:
            print(row)

    elif choice == 3:
        cur.execute("SELECT * FROM bookings")
        data = cur.fetchall()

        for row in data:
            print(row)

    elif choice == 4:
        name = input("Enter car name: ")

        cur.execute(
            "SELECT * FROM cars WHERE car_name = %s",
            (name,)
        )

        data = cur.fetchall()

        for row in data:
            print(row)

        if len(data) == 0:
            print("Car not found.")
        
    elif choice == 5:
        booking_id = input("Enter booking ID: ")
        car_id = input("Enter car ID: ")
        customer_id = input("Enter customer ID: ")
        days = int(input("Enter number of days: "))

        cur.execute(
            "SELECT rent_per_day, status FROM cars WHERE car_id = %s",
            (car_id,)
        )

        car = cur.fetchone()

        if car is None:
            print("Car not found.")

        elif car[1] != "Available":
            print("Car is not available.")

        else:
            rent = car[0]
            total = rent * days

            pickup_date = input("Enter booking date (YYYY-MM-DD): ")
            pickup_time = input("Enter pickup time (HH:MM:SS): ")

            from datetime import datetime, timedelta

            date_obj = datetime.strptime(pickup_date, "%Y-%m-%d")
            return_date = date_obj + timedelta(days=days)

            cur.execute(
                """INSERT INTO bookings
                (booking_id, customer_id, car_id, pickup_date,
                pickup_time, return_date, rental_days,
                total_amount, booking_status)
                VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s)""",
                (booking_id, customer_id, car_id, pickup_date,
                 pickup_time, return_date.date(), days,
                 total, "Ongoing")
            )

            cur.execute(
                """UPDATE cars
                SET status = 'Booked',
                    Availability = 'Booked'
                WHERE car_id = %s""",
                (car_id,)
            )

            con.commit()

            print("Car booked successfully!")
            print("Total amount: Rs.", total)


    elif choice == 6:
        print("Thank you for using Car Rental Management System.")
        break

    else:
        print("Invalid choice. Try again.")

con.close()

```
